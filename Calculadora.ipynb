```python
print("Calculadora de precificação")
```

```python
custos_fixos = {
    "embalagem": 1.50,
    "adesivo": 0.40,
    "fita": 0.30,

}

custo_fixo_total = custos_fixos["embalagem"] + custos_fixos["adesivo"] + custos_fixos["fita"]

print(f"Custo fixo total: R$ {custo_fixo_total:.2f}")
```

```python
preco_peca = float(input("Digite o valor pago pela peça: "))


comissao = preco_peca * 0.40

print("Comissão:", comissao)

```

```python
custo_total = custo_fixo_total + preco_peca + comissao

print(f"Custo total: R$ {custo_total:.2f}")
```

```python
margem_lucro = float(input("Digite a porcentagem de lucro desejada: "))

if margem_lucro < 0:
    print("A margem de lucro não pode ser negativa.")
elif margem_lucro < 10:
    print("A margem mínima de lucro é 10%.")
else:
    preco_de_venda = custo_total + (custo_total * (margem_lucro / 100))
    lucro = preco_de_venda - custo_total

    print(f"Margem aplicada: {margem_lucro:.0f}%")
    print(f"Valor de venda: R$ {preco_de_venda:.2f}")
    print(f"Lucro obtido: R$ {lucro:.2f}")
```

```python
lucro = preco_de_venda - custo_total

print(f"Lucro obtido: R$ {lucro:.2f}")
```
